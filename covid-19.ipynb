```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.duplicated().sum()

df.drop_duplicates(inplace=True)
```

```python
df["continent"] = df["continent"].fillna(0)
df["population"] = df["population"].fillna(0)
df["Recovered"] = df["Recovered"].fillna(0)
df["Deaths"] = df["Deaths"].fillna(0)
df["Tests"]=df["Tests"].fillna(0)
```

```python
df.isnull().sum()
```

```python
df["Recovery_rate"]=(df["Recovered"]/df["Cases"])*100
```

```python
print(df["Recovery_rate"])
```

```python
df["Death_rate"]=(df["Deaths"]/df["Cases"])*100
```

```python
print(df["Death_rate"])
```

```python
df["Deaths_per_Million"] = (df["Deaths"] / df["population"]) * 1000000
```

```python
df["Deaths_per_Million"] = (df["Deaths"] / df["population"]) * 1000000
```

```python
df["Cases_per_100K"] = (df["Cases"] / df["population"]) * 100000
```

```python
df["Tests_per_Person"] = df["Tests"] / df["population"]
```
